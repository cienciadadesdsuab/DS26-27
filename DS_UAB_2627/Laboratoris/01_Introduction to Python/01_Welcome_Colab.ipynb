## Google Colab Specific Tips

### Working with Files

Unlike a local Jupyter notebook, Google Colab notebooks run on Google's servers. Here are the ways to work with files:

1. **Upload files**: Click on the folder icon (📁) on the left sidebar, then click the upload icon
2. **Google Drive**: Mount your Google Drive to access files stored there (see cell below)
3. **Download results**: Right-click on files in the file explorer and select "Download"

### Key Differences from Local Jupyter

1. **No need to save**: Your work is automatically saved to Google Drive
2. **Runtime resets**: If you don't use the notebook for a while, the runtime will disconnect. Just run a cell again to reconnect
3. **File system**: Files are stored in `/content/` (temporary) or `/content/drive/` (if you mount Google Drive)
4. **Package management**: Most scientific packages (numpy, matplotlib, pandas, sklearn) are pre-installed
5. **Permissions**: Click the "Share" button (top right) to share your notebook with others
