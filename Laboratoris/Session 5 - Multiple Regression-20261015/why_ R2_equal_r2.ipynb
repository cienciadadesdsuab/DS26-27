# The Proof: Why $\text{R}^2 = r^2$ in Simple Linear Regression

The proof involves manipulating the components of the $\text{R}^2$ formula, specifically using the fact that the **slope** ($\hat{\beta}_1$) of the simple regression line is directly related to the correlation $r_{xy}$.

---

## 1. The Slope's Formula

In simple linear regression ($\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 x$), the estimated slope $\hat{\beta}_1$ is defined by:

$$\hat{\beta}_1 = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}$$

This can be rewritten using the **correlation coefficient ($r_{xy}$)** and the **standard deviations ($s_x, s_y$)**:

$$\hat{\beta}_1 = r_{xy} \frac{s_y}{s_x}$$

---

## 2. Relating Explained Variation ($\text{SSR}$) to $r_{xy}$

$\text{R}^2$ can also be expressed as the ratio of **Explained Variation ($\text{SSR}$)** to **Total Variation ($\text{SST}$)**:

$$\text{R}^2 = \frac{\text{SSR}}{\text{SST}} = \frac{\sum (\hat{y}_i - \bar{y})^2}{\sum (y_i - \bar{y})^2}$$

We substitute the prediction formula for $\hat{y}_i$, which is derived from the OLS line passing through the means: $\hat{y}_i = \bar{y} + \hat{\beta}_1(x_i - \bar{x})$.

$$
\begin{align*}
\text{SSR} &= \sum [\bar{y} + \hat{\beta}_1(x_i - \bar{x}) - \bar{y}]^2 \\
\text{SSR} &= \sum [\hat{\beta}_1(x_i - \bar{x})]^2 \\
\text{SSR} &= (\hat{\beta}_1)^2 \sum (x_i - \bar{x})^2
\end{align*}
$$

---

## 3. Substituting the Slope $\hat{\beta}_1$ with $r_{xy}$

Now, we substitute the correlation-based expression for the slope ($\hat{\beta}_1 = r_{xy} \frac{s_y}{s_x}$) into the $\text{SSR}$ formula:

$$\text{SSR} = \left( r_{xy} \frac{s_y}{s_x} \right)^2 \sum (x_i - \bar{x})^2$$

Since $\sum (x_i - \bar{x})^2$ is proportional to $s_x^2$ (specifically, $\text{SST}_x = (n-1)s_x^2$), and $\sum (y_i - \bar{y})^2$ is $\text{SST}_y = (n-1)s_y^2$, we can simplify this expression to show that:

$$\text{SSR} = (r_{xy})^2 \times \sum (y_i - \bar{y})^2$$

---

## 4. Final Proof Step

Finally, we substitute this new expression for $\text{SSR}$ back into the $\text{R}^2$ formula:

$$\text{R}^2 = \frac{\text{SSR}}{\text{SST}} = \frac{(r_{xy})^2 \times \sum (y_i - \bar{y})^2}{\sum (y_i - \bar{y})^2}$$

The terms $\sum (y_i - \bar{y})^2$ (which is $\text{SST}$) cancel out, leaving the identity:

$$\mathbf{R^2 = (r_{xy})^2}$$
