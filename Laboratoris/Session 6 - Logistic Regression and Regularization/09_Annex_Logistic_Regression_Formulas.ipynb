## 3. Log Loss (Negative Log-Likelihood): $$J(\hat{y}, y) = -\left(y \log(\hat{y}) + (1-y) \log(1-\hat{y})\right)$$

**Negative Log-Likelihood (NLL)** is the standard error metric for classification models whose output is a probability  

**1. Cost when True Label $y = 1$**

The cost formula is: $\text{Cost} = -\log(p)$

**2. Cost when True Label $y = 0$**

The cost formula is: $\text{Cost} = -\log(1-p)$

**Compact Log Loss (Negative Log-Likelihood) Table**

This table illustrates the cost incurred by a model's predicted probability ($\mathbf{p}$) for both possible true outcomes ($y=1$ and $y=0$). The cost is calculated using the Natural Logarithm ($\ln$).

| Predicted Probability ($\mathbf{p}$) | Cost when $\mathbf{y = 1} $  ($-\ln(p)$) | Cost when $\mathbf{y = 0} $  ($-\ln(1-p)$) | Interpretation |
| :---: | :---: | :---: | :--- |
| **1.0** (Perfect) | $\mathbf{0}$ | $\infty$ | **Perfect Guess** when $y=1$; **Worst Possible Error** when $y=0$. |
| **0.99** (Very Confident) | $\approx \mathbf{0.01}$ | $\approx 4.61$ | Very low error for $y=1$; extremely high error for $y=0$. |
| **0.5** (Uncertain) | $\approx 0.693$ | $\approx 0.693$ | Moderate, equal error for both cases. |
| **0.01** (Very Confident) | $\approx 4.61$ | $\approx \mathbf{0.01}$ | Extremely high error for $y=1$; very low error for $y=0$. |
| **0.0** (Perfect) | $\infty$ | $\mathbf{0}$ | **Worst Possible Error** when $y=1$; **Perfect Guess** when $y=0$. |

## 4.  Derivation of the Logistic Regression Gradient (Log Loss)

This is the mathematical simplification of the partial derivative of the Negative Log-Likelihood (Log Loss) function with respect to a single weight $\beta_j$.

The final result is the formula used for Gradient Descent:
$$\frac{\partial \text{Loss}}{\partial \beta_j} = \mathbf{(p - y) \cdot x_j}$$

---

**1. Key Function Definitions**

We define the primary components involved: 

| Function | Notation | Formula|
 :---: | :---: | :---|
| **Linear Combination (Logit)** | $z$ | $z = \mathbf{x} \cdot \mathbf{\beta} = \sum_{k=0}^n \beta_k x_k$|
| **Predicted Probability (Sigmoid)** | $p$ | $p = \text{logistic}(z) = \frac{1}{1 + e^{-z}}$|
| **Loss Function (Binary Cross-Entropy)** | $\text{Loss}$ | $\text{Loss} = -[y \ln(p) + (1-y) \ln(1-p)]$|
    
---
**2. The Crucial Derivative Property**

The first crucial step involves the derivative of the Sigmoid function with respect to its input $z$. This derivative simplifies to a remarkable form:

$$\frac{dp}{dz} = \frac{d}{dz} \left[ \frac{1}{1 + e^{-z}} \right] = \mathbf{p(1-p)}$$

---

**3. Applying the Chain Rule**

We want to find the partial derivative of the Loss with respect to the $j$-th weight, $\beta_j$. We use the Chain Rule, as the Loss depends on $p$, which depends on $z$, which depends on $\beta_j$:

$$\frac{\partial \text{Loss}}{\partial \beta_j} = \left( \frac{d\text{Loss}}{dp} \cdot \frac{dp}{dz} \right) \cdot \frac{\partial z}{\partial \beta_j}$$

---

**4. Step-by-Step Derivations**

**A. Derivative of Loss with respect to $p$ ($\frac{d\text{Loss}}{dp}$)**

This step introduces a fraction that will later cancel the $p(1-p)$ term.

$$\frac{d\text{Loss}}{dp} = -\left[ y \cdot \frac{1}{p} + (1-y) \cdot \frac{1}{1-p} \cdot (-1) \right]$$

$$\frac{d\text{Loss}}{dp} = - \left[ \frac{y}{p} - \frac{1-y}{1-p} \right]$$

Finding a common denominator $p(1-p)$ and simplifying the numerator:
$$\frac{d\text{Loss}}{dp} = - \left[ \frac{y(1-p) - p(1-y)}{p(1-p)} \right] = - \left[ \frac{y - yp - p + yp}{p(1-p)} \right]$$

$$\frac{d\text{Loss}}{dp} = - \left[ \frac{y - p}{p(1-p)} \right] = \mathbf{\frac{p - y}{p(1-p)}}$$

---

**B. Derivative of $z$ with respect to $\beta_j$ ($\frac{\partial z}{\partial \beta_j}$)**

The derivative of the linear sum $\mathbf{x} \cdot \mathbf{\beta}$ with respect to one specific weight $\beta_j$ is simply the feature associated with that weight, $x_j$.

$$\frac{\partial z}{\partial \beta_j} = \frac{\partial}{\partial \beta_j} (\beta_0 x_0 + \dots + \beta_j x_j + \dots) = \mathbf{x_j}$$

---

**5. The Final Simplification (The Cancellation)**

Substitute all derived terms back into the Chain Rule equation:

$$\frac{\partial \text{Loss}}{\partial \beta_j} = \left( \mathbf{\frac{p - y}{p(1-p)}} \right) \cdot \left( \mathbf{p(1-p)} \right) \cdot \left( \mathbf{x_j} \right)$$

The terms $\mathbf{p(1-p)}$ and $\frac{1}{\mathbf{p(1-p)}}$ **cancel out completely**.

$$\frac{\partial \text{Loss}}{\partial \beta_j} = \mathbf{(p - y) \cdot x_j}$$

This formula means the gradient—the signal for adjusting the weight $\beta_j$—is simply the **error in prediction** multiplied by the **value of the feature $x_j$**.
